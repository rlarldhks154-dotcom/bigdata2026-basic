### 1. f.read() 방식과 with open() 방식 두 가지로 작성

```python
with open('data1.txt', 'w', encoding='utf-8') as f:
    f.write("Life is too short.\n")
    f.write("So, you need Python.\n")
    f.write("Python is easy,\n")
    f.write("and useful on many fields.\n")
```

```python
f = open('data1.txt', 'r', encoding='utf-8')
data = f.read()
print(data)
f.close()
```

```python
with open('data1.txt', 'r', encoding='utf-8') as f:
    data = f.read()
    print(data)
```

### 2. 과 졸업 전시회 참석 학생 이름을 입력받아 'welcome!!'을 출력하고, 명단을 guest.txt에 저장한 후 총 인원 수를 출력하세요.

```python
count = 0

with open('guest.txt', 'w', encoding='utf-8') as f:
    while True:
        name = input("참석 학생 이름을 입력하세요 (종료하려면 'q' 입력): ")
        if name == '0':
            break
        
        print(f"{name}님 welcome!!")
        f.write(name + '\n')  # 입력받은 이름을 파일에 기록
        count += 1

print(f"총 참석 인원 수: {count}명 (guest.txt 파일에 저장되었습니다.)")
```

### 3. data2.txt에 저장된 점수를 모두 읽어 합계와 평균을 구한 후 화면에 출력하세요.

```python
with open('data2.txt', 'w', encoding='utf-8') as f:
    print("점수를 하나씩 입력하세요. (입력을 끝내려면 'q'를 입력)")
    while True:
        user_input = input("점수 입력: ")
        
        if user_input.lower() == 'q':  # q나 Q를 입력하면 종료
            break

        f.write(user_input + '\n')
```

```python
total = 0
count = 0

with open('data2.txt', 'r', encoding='utf-8') as f:
    for line in f:
        score = int(line.rstrip())
        total += score
        count += 1

if count > 0:
    average = total / count
    print(f"읽은 데이터 개수: {count}개")
    print(f"합계: {total}")
    print(f"평균: {average:.2f}")
else:
    print("입력된 점수 데이터가 없습니다.")
```

### 4. data3.txt에 저장된 신조어 단어들을 공백 기준으로 분리하여 한 단어씩 출력하세요.

```python
with open('data3.txt', 'w', encoding='utf-8') as f:
    print("신조어 단어를 하나씩 입력하세요. (입력을 끝내려면 'z'를 입력)")
    while True:
        word_input = input("단어 입력: ")
        
        if word_input.lower() == 'z':
            break
        
        f.write(word_input + ' ')
```

```python
with open('data3.txt', 'r', encoding='utf-8') as f:
    content = f.read()
    words = content.split()

for word in words:
    print(word)
```

### 5.MT 장소 투표 프로그램 — 딕셔너리(가령, 대성리, 남이섬, 청평)에 입력받은 장소의 득표수를 누적하여 출력하세요.

```python
vote_result = {'가평': 0, '대성리': 0, '남이섬': 0, '청평': 0}

while True:
    place = input("투표할 장소를 입력하세요 (가평/대성리/남이섬/청평, 종료는 '결과'): ")
    
    if place == '결과':
        break
    
    if place in vote_result:
        vote_result[place] += 1
        print(f"[{place}]에 투표")
    else:
        print("후보에 없는 장소")

for key, value in vote_result.items():
    print(f"{key}: {value}표")
```
