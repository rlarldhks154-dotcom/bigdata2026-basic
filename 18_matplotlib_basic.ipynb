# Matplotlib 라이브러리 기본 시각화

```python
# 라이브러리 불러오기
import matplotlib.pyplot as plt
```

```python
# 데이터
xdata = [2, 3, 4, 5]
ydata = [1, 3, 5, 7]
plt.plot(xdata, ydata) # 선 그래프 (라인 그래프, 꺾은선 그래프)
plt.title('Line Graph')
plt.show()
```

## 범례 --> plt.legend()

```python
data1 = [2, 3, 4, 5]
data2 = [8, 6, 4, 2]
plt.plot(data1, label='오름차순') # asc = 오름차순
plt.plot(data2, label='내림차순') # desc = 내림차순
plt.legend() # 범례
plt.show()
```

## 선에 색깔 넣기

```python
plt.plot(xdata, ydata, color='red')
plt.show()
```

## X축 제목, Y축 제목
- plt.xlabel()
- plt.ylabel()

```python
plt.plot(xdata, ydata, color='k')
plt.xlabel('X축 제목')
plt.ylabel('Y축 제목')
plt.show()
```

## 선 종류

- linestyle = 'dash' 또는 linestyle = '--'
- 일반 실선 = linestyle = 'solid' 또는 linestyle = '-'

```python
plt.plot(data1, color='r', label='dashed', linestyle='--') # 대시 긴 점선 --> 파선(쇄선)
plt.plot(data2, color='g', label='dotted', linestyle=':') # 점선
plt.plot(ydata, color='b', label='Dash-dot', linestyle='-.') # 1점 쇄선
plt.legend()
plt.show()
```

## 축 범위 조정

- plt.xlim(범위)
- plt.ylim(범위)

```python
plt.title('X, Y range')
xdata = [5, 10, 15, 20]
ydata = [1, 3, 5, 7]
plt.plot(xdata, ydata, color='b', linestyle='--', marker='o', markersize='5')
plt.xlim(0, 25)
plt.ylim(-2, 10)
plt.show()
```
