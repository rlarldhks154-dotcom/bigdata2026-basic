```python
menu_list = [
    ["에스프레소", 2000, "커피"],
    ["아메리카노", 2500, "커피"],
    ["카푸치노", 3500, "커피"],
    ["카페라떼", 3500, "커피"],
    ["모카라떼", 4000, "커피"],
    ["바닐라라떼", 4000, "커피"],
    ["소이라떼", 4000, "커피"],
    ["피넛라떼", 4000, "커피"],
    ["토피넛라떼", 4200, "커피"],
    ["화이트모카", 4200, "커피"],
    ["카라멜마끼아또", 4500, "커피"],

    ["프라프치노", 4800, "음료"],
    ["핫초코", 3500, "음료"],
    ["레몬에이드", 4000, "음료"],
    ["청포도에이드", 4000, "음료"],
    ["자몽에이드", 4000, "음료"],
    ["스무디", 4500, "음료"],
    ["망고스무디", 4800, "음료"],
    ["딸기스무디", 4800, "음료"],

    ["초코쿠키", 2000, "디저트"],
    ["화이트쿠키", 2000, "디저트"],
    ["피넛쿠키", 2000, "디저트"],
    ["당근케이크", 5000, "디저트"],
    ["초코케이크", 5000, "디저트"],
    ["치즈케이크", 5000, "디저트"]
]

all_menu_names = [menu[0] for menu in menu_list]
print(" ".join(all_menu_names))
print()

while True:
    print("1.가격별 메뉴 조회 2.종류별 메뉴 조회 3.주문 0.종료 : ", end="")
    try:
        choice = int(input())
    except ValueError:
        print("올바른 숫자를 입력해주세요.\n")
        continue
        
    if choice == 0:
        print("스마트 카페 메뉴조회 시스템 종료!")
        break
        
    elif choice == 1:
        print("최저가격, 최고가격 입력(예:1000,2000) ", end="")
        try:
            low, high = map(int, input().split(','))
        except ValueError:
            print("형식에 맞춰 올바르게 입력해주세요. (예: 2000,2500)\n")
            continue
            
        result = []
        for menu in menu_list:
            name, price, category = menu
            if low <= price <= high:
                result.append([name, price])
                
        print("\n< 입력조건의 메뉴 목록 >")
        print(result)
        print()
        
    elif choice == 2:
        print("종류별 메뉴 입력(예:커피,음료,디저트) ", end="")
        category_input = input().strip()
        
        result = []
        for menu in menu_list:
            name, price, category = menu
            if category == category_input:
                result.append([name, price])
                
        print("\n< 입력조건의 메뉴 목록 >")
        print(result)
        print()
        
    elif choice == 3:
        print("원하는 메뉴 입력 : ", end="")
        menu_input = input().strip()
        
        result = []
        for menu in menu_list:
            name, price, category = menu
            if name == menu_input:
                result.append([name, price, category])
                
        print("\n< 입력조건의 메뉴 목록 >")
        print(result)
        print()
        
    else:
        print("메뉴에 없는 번호입니다. 다시 선택해주세요.\n")
```
